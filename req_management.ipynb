# Тема 2.1. Управление требованиями: от сбора требований до Job Stories

## 🎯 Цели занятия

После этого занятия вы сможете:

1.  Собирать требования из разных источников и работать со стейкхолдерами.
2.  Анализировать и классифицировать требования (FR / NFR), приоритизировать их.
3.  Составлять техническое задание (ТЗ) по классической и гибкой модели.
4.  Формулировать User Stories по критериям INVEST.
5.  Объяснять подход **Jobs to be Done (JTBD)** и применять его на практике.
6.  Писать **Job Stories** по каноническому шаблону.

## 📚 Часть 1. Сбор требований к ПО

### 1.1. Что такое требование?

**Требование** — это описание того, что система должна делать, какими свойствами обладать и в каких условиях работать.

### 1.2. Источники требований

| Источник | Что даёт |
|----------|----------|
| Заказчик | Бизнес-цели, бюджет, сроки |
| Конечные пользователи | Реальные сценарии использования |
| Регуляторы | Юридические и отраслевые нормы |
| Внутренние стейкхолдеры | Технические ограничения, поддержка |
| Аналитика | Метрики, поведение пользователей |

### 1.3. Методы сбора требований

#### 🎤 Интервью

**Что это**: структурированная или полуструктурированная беседа с одним или несколькими стейкхолдерами для выявления их потребностей и ожиданий.

**Когда применять**: на ранних этапах проекта, когда нужно глубоко понять проблему, бизнес-контекст и скрытые ожидания заказчика. Особенно эффективно, когда проект не имеет строгих временных ограничений, а стейкхолдеры доступны для диалога[reference:0].

**Особенности**:
- Позволяет задавать уточняющие вопросы и «копать» глубже.
- Даёт в основном качественные данные, но может содержать и количественные[reference:1].
- Требует от интервьюера навыков активного слушания и умения направлять беседу.

**Техники**:
- **«5 почему»** — итеративное углубление в причину проблемы.
- **Открытые вопросы** («Что для вас важно в этом процессе?») — для сбора контекста.
- **Закрытые вопросы** («Должна ли быть возможность экспорта в PDF?») — для уточнения деталей.
- **Техника «Расскажите мне о последнем случае...»** — помогает перейти от общих пожеланий к конкретным сценариям.

#### 📝 Анкетирование (опросы)

**Что это**: сбор требований через структурированные опросники, которые рассылаются большому числу респондентов.

**Когда применять**: когда нужно охватить большое количество пользователей с минимальными ресурсами. Эффективно для сбора количественных данных и статистики[reference:2].

**Особенности**:
- Позволяет собрать данные от многих людей одновременно.
- Качество результата критически зависит от дизайна опросника[reference:3].
- Не позволяет задавать уточняющие вопросы «на лету».

**Техники**:
- Использование шкал Лайкерта (от «категорически не согласен» до «полностью согласен»).
- Комбинирование закрытых (для статистики) и открытых (для инсайтов) вопросов.
- Пилотное тестирование опросника на малой группе перед рассылкой.

#### 👁️ Наблюдение

**Что это**: метод, при котором аналитик наблюдает за реальной работой пользователей в их естественной среде, не вмешиваясь в процесс.

**Когда применять**: когда слова пользователей расходятся с их действиями. Наблюдение позволяет выявить неосознанные потребности и «теневые» процессы, о которых пользователи не рассказывают в интервью.

**Особенности**:
- Даёт наиболее достоверные данные о реальных сценариях работы.
- Может быть трудоёмким и требовать длительного присутствия.
- Существует эффект наблюдателя (люди ведут себя иначе, когда за ними смотрят).

**Техники**:
- **Полевое наблюдение** — в реальных условиях.
- **Apprenticing (ученичество)** — аналитик сам выполняет работу пользователя под его руководством.
- **Контекстное интервью** — сочетание наблюдения и беседы в момент выполнения задачи.

#### 📄 Анализ документов

**Что это**: изучение существующей документации: регламентов, инструкций, старых ТЗ, логов, схем бизнес-процессов.

**Когда применять**: на старте проекта, когда есть накопленная документация. Особенно полезно для проектов модернизации или замены существующих систем.

**Особенности**:
- Позволяет быстро получить представление о предметной области.
- Документы могут быть устаревшими или неполными.
- Требует навыков анализа и сопоставления информации из разных источников.

**Техники**:
- **Системная археология** — восстановление требований по существующей системе.
- **Анализ пользовательских интерфейсов** — изучение существующих экранов и сценариев.
- **Составление глоссария** — единый словарь терминов предметной области.

#### 🧠 Воркшопы и фокус-группы

**Что это**: групповые сессии, в которых стейкхолдеры совместно обсуждают и формулируют требования.

**Когда применять**: когда нужно собрать множество точек зрения и достичь консенсуса по спорным вопросам. Эффективно для выявления областей согласия и конфликта[reference:4].

**Особенности**:
- Позволяет собрать множество мнений за короткое время.
- Выявляет зоны консенсуса и разногласий.
- Требует опытного фасилитатора для управления групповой динамикой.

**Техники**:
- **JAD (Joint Application Design)** — совместная разработка требований с участием всех сторон.
- **Мозговой штурм** — генерация идей без критики.
- **Ролевые игры** — проигрывание сценариев использования будущей системы.

#### 🖥️ Прототипирование

**Что это**: создание упрощённых моделей будущего продукта (от бумажных макетов до интерактивных прототипов) для проверки гипотез и уточнения требований.

**Когда применять**: когда словесное описание недостаточно и нужно «показать» будущий продукт. Особенно эффективно для сложных интерфейсов и инновационных решений.

**Особенности**:
- Позволяет получить обратную связь на раннем этапе.
- Снижает риск неверного понимания требований.
- Может быть ресурсоёмким при создании интерактивных прототипов.

**Техники**:
- **Бумажное прототипирование** — быстрые наброски экранов.
- **Wireframes** — схематичные макеты без дизайна.
- **Интерактивные прототипы** — кликабельные модели (Figma, Axure).
- **MVP (Minimum Viable Product)** — минимально жизнеспособный продукт для проверки гипотез.

### 1.4. Карта стейкхолдеров (Stakeholder Map)

**Что это**: визуальный инструмент для идентификации всех заинтересованных сторон проекта и анализа их влияния и интереса.

**Зачем нужна**:
1.  **Идентификация** — убедиться, что все заинтересованные стороны учтены.
2.  **Приоритизация** — понять, с кем нужно работать плотнее, а кого достаточно информировать.
3.  **Управление ожиданиями** — выявить потенциальные конфликты интересов на раннем этапе.
4.  **Планирование коммуникации** — определить стратегию взаимодействия с каждой группой.

**Матрица «Влияние × Интерес»**:

| | Низкий интерес | Высокий интерес |
|---|---|---|
| **Высокое влияние** | Держать довольными | Управлять плотно |
| **Низкое влияние** | Мониторить | Информировать |

**Стратегии управления стейкхолдерами**:

- **Управлять плотно** (высокое влияние, высокий интерес): регулярные встречи, вовлечение в принятие решений, совместная работа над требованиями. Это ключевые стейкхолдеры, которые могут как помочь, так и заблокировать проект[reference:5].
- **Держать довольными** (высокое влияние, низкий интерес): периодические отчёты, краткие обновления. Они не хотят вникать в детали, но их поддержка важна.
- **Информировать** (низкое влияние, высокий интерес): рассылки, доступ к документации. Они заинтересованы, но не принимают ключевых решений.
- **Мониторить** (низкое влияние, низкий интерес): минимальное взаимодействие, но следить за изменениями.

**Управление ожиданиями стейкхолдеров**:

Управление ожиданиями — это непрерывный процесс, который включает:
1.  **Раннее выявление ожиданий** — на этапе сбора требований выяснить, что каждый стейкхолдер ожидает от проекта.
2.  **Согласование реалистичных целей** — если ожидания завышены, честно объяснить ограничения и предложить альтернативы.
3.  **Регулярная коммуникация** — держать стейкхолдеров в курсе прогресса, рисков и изменений.
4.  **Управление изменениями** — если требования меняются, оценить влияние на сроки и бюджет и согласовать изменения со всеми затронутыми сторонами.
5.  **Документирование договорённостей** — все ключевые решения фиксируются в ТЗ, протоколах встреч или backlog.

> 💡 **Совет**: матрица «власть/интерес» может быть модифицирована для более точного отражения характера интереса стейкхолдеров (например, размерность 2×4)[reference:6].

## 📚 Часть 2. Анализ требований

### 2.1. Классификация

- **Бизнес (BR)** — *что* ожидает от системы бизнес. Пример: «Увеличить количество продаж онлайн на 20%».
- **Пользовательские (UR)** — *что* должен делать пользователь в системе. Пример: «Пользователь хочет получить список "горячих" клиентов для рассылки предложения о покупке».
- **Функциональные (FR)** — *что* делает система. Пример: «Пользователь может восстановить пароль по email».
- **Нефункциональные (NFR)** — *как* система это делает. Пример: «Отклик < 200 мс», «Доступность 99.9%», «Соответствие GDPR».

**Основные категории нефункциональных требований**:

| Категория | Описание | Пример |
|-----------|----------|--------|
| **Производительность** | Время отклика, пропускная способность | «Загрузка страницы < 2 сек» |
| **Надёжность** | Доступность, отказоустойчивость | «Доступность 99.9%» |
| **Безопасность** | Аутентификация, шифрование, аудит | «Данные шифруются по AES-256» |
| **Удобство использования** | Юзабилити, доступность | «Интерфейс соответствует WCAG 2.1» |
| **Расширяемость** | Возможность добавления функций | «Поддержка плагинов» |
| **Сопровождаемость** | Простота внесения изменений | «Покрытие тестами > 80%» |
| **Переносимость** | Работа на разных платформах | «Поддержка Windows, macOS, Linux» |

Нефункциональные требования могут быть классифицированы как продуктовые, организационные и внешние[reference:7].

### 2.2. Приоритизация

- **MoSCoW**: Must / Should / Could / Won't.
- **Метод Кано**: обязательные, одномерные, привлекательные.
- **RICE**: Reach × Impact × Confidence / Effort.

### 2.3. Проверка качества требований

Требование должно быть: **полным, непротиворечивым, реализуемым, тестируемым, однозначным, атомарным**[reference:8].

**Критерии качества**:
1.  **Корректность** — требование действительно отражает потребность.
2.  **Недвусмысленность** — только одно толкование.
3.  **Полнота** — все необходимые детали указаны[reference:9].
4.  **Непротиворечивость** — требования не конфликтуют друг с другом[reference:10].
5.  **Тестируемость** — можно проверить выполнение требования[reference:11].
6.  **Реализуемость** — технически осуществимо.
7.  **Атомарность** — одно требование = одна потребность.

> ⚠️ **Типичные ошибки**: «додумывание» за заказчика, размытые формулировки («удобный интерфейс»), конфликтующие требования, требования-решения («нужна кнопка») вместо требований-проблем («пользователь должен быстро оформить заказ»).

## 📚 Часть 3. Техническое задание (ТЗ)

### 3.1. Что такое ТЗ

**ТЗ** — документ, фиксирующий требования к системе, условия её создания и критерии приёмки. В классической модели — основа договора между заказчиком и исполнителем.

### 3.2. Структура ТЗ по ГОСТ 34.602-89 (сокращённо)

Согласно ГОСТ 34.602-89, ТЗ на автоматизированную систему содержит следующие разделы[reference:12]:

1.  **Общие сведения** — наименование системы, заказчик, исполнитель, основание для разработки.
2.  **Назначение и цели создания (развития) системы** — для чего создаётся система, какие цели преследует.
3.  **Характеристика объектов автоматизации** — описание предметной области.
4.  **Требования к системе** — функциональные и нефункциональные требования, требования к видам обеспечения[reference:13].
5.  **Состав и содержание работ по созданию (развитию) системы** — перечень стадий и этапов работ, сроки выполнения, исполнители[reference:14].
6.  **Порядок контроля и приёмки** — как будет проверяться соответствие системы требованиям.
7.  **Требования к документированию** — какие документы должны быть созданы.
8.  **Источники разработки** — нормативные документы, стандарты, руководства.

### 3.3. Современная эволюция: ТЗ → PRD → Vision

| Документ | Когда используется |
|----------|--------------------|
| **ТЗ (ГОСТ)** | Госзаказ, крупные интеграции, жёсткие контракты |
| **PRD (Product Requirements Document)** | Продуктовая разработка, гибкие команды |
| **Vision & Scope** | Старт проекта, формулировка видения |
| **Живая документация** | Agile: требования = backlog + User Stories |

## 📚 Часть 4. User Stories: классический Agile-формат

### 4.1. Шаблон

```
As a [role], I want to [action], so that [benefit].
```

### 4.2. Критерии качества: INVEST

**INVEST** — это акроним, описывающий характеристики качественной User Story[reference:15]:

- **I**ndependent (Независимая) — story не зависит от других story, что упрощает планирование и приоритизацию[reference:16].
- **N**egotiable (Обсуждаемая) — story — это не контракт, а предмет для обсуждения с командой и заказчиком[reference:17].
- **V**aluable (Ценная) — story приносит ценность пользователю или заказчику.
- **E**stimable (Оцениваемая) — команда может оценить трудоёмкость story.
- **S**mall (Маленькая) — story достаточно мала, чтобы поместиться в одну итерацию[reference:18].
- **T**estable (Тестируемая) — можно проверить, выполнена ли story.

### 4.3. Definition of Ready / Done

- **DoR** (Definition of Ready): story понятна команде, есть критерии приёмки, оценена, не имеет блокирующих зависимостей.
- **DoD** (Definition of Done): код написан, протестирован, задеплоен, документация обновлена, критерии приёмки выполнены.

### 4.4. Ограничения User Stories

- **Роль слишком широкая** — «as a user» не даёт контекста. Кто этот пользователь? В какой ситуации он находится?
- **Фокус на *кто*, а не на *когда* и *почему*** — теряется контекст использования.
- **Скатывание в «список фич»** — story превращается в перечень функций без понимания ценности.
- **Игнорирование ситуации** — одна и та же роль в разных ситуациях может иметь разные потребности.

## 📚 Часть 5. Jobs to be Done (JTBD)

### 5.1. Что такое JTBD?

**Jobs to be Done** — теория поведения потребителей: люди не покупают продукты, а «нанимают» их для выполнения «работы» в конкретном контексте.

> «People don't buy products — they hire them to make progress in their lives». — К. Кристенсен[reference:19]

### 5.2. Классический кейс «Молочный коктейль»

Сеть фастфуда обнаружила: большинство коктейлей продаётся до 8:30 утра, покупатели пьют их за рулём по дороге на работу. «Работа» коктейля — скрасить скучную поездку и не проголодаться до обеда.

### 5.3. Три ключевых принципа

1.  **Задача важнее продукта** — пользователь выбирает продукт ради результата, а не ради характеристик.
2.  **Контекст важнее демографии** — возраст, пол и другие демографические признаки не предсказывают поведение. Важна ситуация, в которой находится человек.
3.  **Прогресс важнее характеристик** — люди «нанимают» продукты, чтобы продвинуться в желаемом состоянии.

### 5.4. Switch-интервью и «четыре силы»

**Switch-интервью** — метод исследования, при котором реконструируется момент, когда клиент решил сменить решение. Интервью фокусируется на **недавних переключателях** (людях, которые недавно начали или перестали пользоваться продуктом)[reference:20].

**Четыре силы**, определяющие решение о смене[reference:21]:

| Сила | Вопрос |
|------|--------|
| **Push** (толчок) | Что не устраивало в текущей ситуации? |
| **Pull** (притяжение) | Что привлекло в новом решении? |
| **Anxiety** (тревога) | Что заставило сомневаться? |
| **Habit** (привычка) | Что удерживало от смены? |

**Формула переключения**: Switching Threshold = (Pushes + Pulls) − (Habits + Anxieties). Если результат положительный — клиент переключается[reference:22].

**Процесс switch-интервью**:
1.  **Отбор участников** — только недавние переключатели[reference:23].
2.  **Построение таймлайна** — от первого сомнения до первого использования нового продукта[reference:24].
3.  **Выявление четырёх сил** — вопросы о push, pull, anxiety, habit[reference:25].
4.  **Определение «работы»** — функциональной, социальной и эмоциональной[reference:26].

## 📚 Часть 6. Job Stories

### 6.1. Шаблон (Alan Klement, 2013)

```
When [situation], I want to [motivation], so I can [outcome].
```

**Пример**:
```
Когда я настраиваю радиатор и хочу получить конкретную мощность нагрева,
я хочу быстро определить подходящую конфигурацию,
чтобы не тратить время на поиск оптимального варианта.
```

### 6.2. User Story vs Job Story

| Критерий | User Story | Job Story |
|----------|-----------|-----------|
| Фокус | Роль (*Who*) | Ситуация (*When*) |
| Контекст | Минимален | Явно указан |
| Мотивация | Вторична | Центральна |
| Результат | Пожелание | Измеримый прогресс |

### 6.3. Алгоритм формулировки

1.  **When** — конкретный триггер, наблюдаемая ситуация. Опишите, что происходит, где и в какой момент.
2.  **I want to** — мотивация без UI-элементов. Что пользователь хочет сделать? Чего он пытается достичь?
3.  **So I can** — измеримый результат. Какой прогресс получит пользователь?

### 6.4. Критерии качества Job Story

- **Контекстность** — ситуация описана конкретно и наблюдаемо.
- **Отсутствие UI-деталей** — не упоминаются кнопки, экраны и т.п.
- **Измеримость результата** — outcome можно проверить.
- **Одна работа на story** — не смешивать несколько потребностей.
